Se crea un portafolio ponderado por precio, es el único que vamos a usar, ambos son muy muy parecidos, solo tiene un poquito más de return el ponderado.

```python
import pandas as pd
```

```python
file = r'cryptos.csv'
data = pd.read_csv(file, index_col="Date", parse_dates=True)
```

# **CREATE AN EQUAL WEIGTHED PORTFOLIO**

```python
equal_weighted_portfolio = data.mean(axis=1)
data['EWP'] = equal_weighted_portfolio
```

# **PONDERATED PORTFOLIO BY PRICE**

```python
cryptos = data.columns[:-1]
```

```python
cryptos
```

```python
totalcost = sum(data.iloc[0, :-1])
totalcost
```

```python
weigths = {}

for crypto in cryptos:
    weigths[crypto] = data.iloc[0][crypto] / totalcost

weigths
```

```python
data['PP'] = 0
for crypto in cryptos:
    data['PP'] += data[crypto] * weigths[crypto]
```

```python
data.to_csv('cryptos-index.csv')
```
