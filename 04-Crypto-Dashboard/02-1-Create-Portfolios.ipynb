**Duración y Justificación del Periodo**

<div style="text-align: justify;">

El periodo de evaluación se extiende desde el 14 de marzo hasta el 16 de abril, fecha que se asocia con el **Día de la Constante de Napier**. John Napier (1550-1617) fue un matemático escocés reconocido por la invención de los logaritmos, una herramienta matemática que revolucionó los cálculos numéricos y facilitó enormemente el trabajo en astronomía, navegación y otras disciplinas científicas.

La elección de este intervalo no es arbitraria, sino que se fundamenta en el simbolismo matemático de ambas fechas. Mientras que $\pi$ es una de las constantes más fundamentales en geometría y análisis matemático, los logaritmos de Napier han sido clave en el desarrollo de métodos computacionales y en la teoría del crecimiento exponencial. La relación entre ambos conceptos puede apreciarse en ecuaciones como:

$$
e^{i\pi} + 1 = 0
$$

donde la identidad de Euler une de manera elegante las constantes $\pi$, $e$ (la base de los logaritmos naturales) y la unidad imaginaria $i$.

```python
import pandas as pd
from pypfopt.efficient_frontier import EfficientFrontier
from pypfopt import risk_models
from pypfopt import expected_returns
```

```python
file = r'cryptos.csv'
data = pd.read_csv(file, index_col="Date", parse_dates=True)
```

```python
mu = expected_returns.mean_historical_return(data, frequency=31)
s = risk_models.sample_cov(data, frequency=31)
```

# **MAX SHARPE BUY ONLY**

```python
efbuy = EfficientFrontier(mu, s)    # Efficient Frontier Buy
```

```python
efbuycopy = efbuy.deepcopy()
```

```python
efbuyms = efbuy.max_sharpe()        # Efficient Frontier Buy Max Sharpe
```

```python
wefbuy = efbuy.clean_weights()      # Weights Efficient Frontier Buy
```

```python
for i in wefbuy:
    if wefbuy[i] > 0:
        print(i, wefbuy[i])
```

```python
efbuy.portfolio_performance(verbose=True)
```

```python
pd.DataFrame(wefbuy, index=[0]).to_csv('portfolio-buy.csv', index=False)
```

# **MAX SHARPE CFDs**

```python
efcfd = EfficientFrontier(mu, s, weight_bounds=(-1, 1))    # Efficient Frontier Buy
```

```python
efcfdcopy = efcfd.deepcopy()        # a copy for not calculating again
```

```python
efcfdms = efcfd.max_sharpe()        # Efficient Frontier Buy Max Sharpe
```

```python
wefcfd = efcfd.clean_weights()      # Weights Efficient Frontier Buy
```

```python
sum = 0
for crypto in wefcfd:
    if wefcfd[crypto] != 0:
        print(crypto, wefcfd[crypto])
        sum = sum + abs(wefcfd[crypto])
```

```python
efcfd.portfolio_performance(verbose=True)
```
