```python
import pandas as pd
from pypfopt import risk_models
from pypfopt import expected_returns
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
file = r'cryptos.csv'
data = pd.read_csv(file, index_col="Date", parse_dates=True)
```

```python
mu = expected_returns.mean_historical_return(data)
s = risk_models.sample_cov(data)
```

```python
cmap = sns.diverging_palette(220, 20, as_cmap=True) # cute palette

g = sns.clustermap(risk_models.cov_to_corr(s), annot=True, cmap=cmap, fmt='.2f', figsize=(14, 14), annot_kws={"size": 8}, cbar_pos=(1.01, 0.08, 0.05, 0.72), vmin=-1, vmax=1)

plt.text(-9, 1.6, "Clustermap de Criptomonedas", fontsize=30, color='black', ha='center')
    
plt.savefig(r'svg/03-clustermap.svg', format='svg', bbox_inches='tight')




plt.show()
```
